```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import joblib
from sklearn.model_selection import train_test_split

PATH = "/content/drive/MyDrive/fraud-detection"
df = pd.read_csv(f"{PATH}/data/paysim.csv")

# 1. النوعين اللي فيهم احتيال بس
df = df[df["type"].isin(["TRANSFER", "CASH_OUT"])].copy()
print("بعد الفلترة:", df.shape)

# 2. Feature Engineering
df["errorBalanceOrig"] = df["newbalanceOrig"] + df["amount"] - df["oldbalanceOrg"]
df["errorBalanceDest"] = df["oldbalanceDest"] + df["amount"] - df["newbalanceDest"]
df["hour"] = df["step"] % 24

# 3. تحويل النوع لرقم (TRANSFER = 1, CASH_OUT = 0)
df["type"] = (df["type"] == "TRANSFER").astype(int)

# 4. شيل الأعمدة اللي ملهاش لازمة
df = df.drop(columns=["nameOrig", "nameDest", "isFlaggedFraud", "step"])

# 5. التقسيم
X = df.drop(columns="isFraud")
y = df["isFraud"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
print("Train:", X_train.shape, "| Test:", X_test.shape)
print("Fraud % Train:", y_train.mean()*100, "| Test:", y_test.mean()*100)

# 6. الحفظ
joblib.dump((X_train, X_test, y_train, y_test), f"{PATH}/data/split.pkl")
print("Saved ✅")
```
